```python
import numpy as np
from scipy.stats import chi2_contingency

#Example: contingency table (observed frequencies)
data = np.array([[50, 30],   #group 1 (e.g., yes/No)
                [20, 40]])

# Perform chi-squard test
chi2, p, dof, expected = chi2_contingency(data)

print("chi-squre Statistic:",dof)
print("P-Value:",p)
print("Degrees of Freedom:",dof)
print("Exected Frequencies:")
print(expected)

if p < 0.05:
  print("Reject the null hypothesis: Variables are dependent.")
else:
  print("Reject the null hypothesis: Variables are independent.")
```

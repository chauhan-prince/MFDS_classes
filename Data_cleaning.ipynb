```python
import pandas as pd
import numpy as np

#Simple data
data = {'A': [1, 2, np.nan, 4],
      'B': [5, np.nan, np.nan, 8],
      'C': ['cat','dog',np.nan,'mouse']}
df = pd.DataFrame(data)
print(df)

#Handle missing values
df['A'].fillna(df['A'].mean(), inplace=True)  # Mean imputation for column A
df['B'].fillna(0, inplace=True)               # Replace with 0 for column B
df['C'].fillna('unknown', inplace=True)       # Replace with 'unkonwn' for column C

print(df)
```

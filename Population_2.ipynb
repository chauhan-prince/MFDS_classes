```python
import numpy as np
from scipy import stats

# give data
population_mean = 400
population_std = 8
sample_size = 100
sample_mean = 398

#calculate the Z- Score
z_score = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))

#calculate the p-value a two- tailedtest
p_value = 2 * (1-stats.norm.cdf(abs(z_score)))

print(f"z-score: {z_score}")
print(f"p-value: {p_value}")

#canclusion
alpha = 0.05 #Significance level
if p_value < alpha:
   print("Reject the null hypothesis: there is a significant difference in coffee can weigth.")
else:
  print("Fail to reject the null hypothesis: no significantr differnce in coffee can weigth.")
```

```python
 # Question 2
from scipy import stats
 #Give data
population_mean = 50       # claimed mean sugar contant
sample_size= 15            # Sample size
sample_mean = 48            # sample standed deviation
sample_std = 5              # Observed sample mean

#Generate a random sample close to the observed sample mean
np.random.seed(0)
sample = np. random.normal(sample_mean, sample_std, sample_size)

# Perform the one-sample t-test
t_statistic , p_value = stats.ttest_1samp(sample, population_mean)

print(f"T-ststistic: {t_statistic}")
print(f"P-value: {p_value}")

#conclusion
alpha = 0.05  #Significance level
if p_value < alpha:
   print("Reject the null hypothesis: there is a significant difference in suger content.")
else:
  print("Fail to reject the null hypothesis: no significantr differnce in suger content.")
```

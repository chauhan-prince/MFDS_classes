```python
import pandas as pd

data = {
    "study Hours" : [5,3,4,6,5,2,3,4,5,6],
    "Recreation Hours": [2,4,3,1,2,5,4,3,2,1],
    "Attendance (%)": [90,75,80,95,85,60,76,75,85,95],
    "Assignment Submitted": [8,5,6,9,7,3,4,6,7,9],
    "Sleep Hours": [7,6,6.5,8,7,5,5.5,6,7,8]
}

df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(pca_result, columns=["PC1","PC2"])
print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)

```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))
plt.scatter(pca_df["PC1"],pca_df["PC2"],c="blue", edgecolor='k', s=100)
plt.title('PCA of Academic Performance Data')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True)
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Age": [25,34,45,23,54,32,40,27,50,33],
    "Income (in $1000s)": [40,60,80,30,120,55,75,50,100,45],
    "Total Spending (in $1000s)": [10,15,20,12,25,18,22,16,28,20],
    "Frequency of Visits": [12,20,30,8,50,18,28,15,35,14],
    "Products Purchased": [5,8,12,3,18,7,11,6,16,7]
}

df = pd.DataFrame(data)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

#Ensuring we get enough components
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(pca_result, columns=["PC1","PC2"])

plt.figure(figsize=(8,6))
plt.scatter(pca_df["PC1"],pca_df["PC2"],c="green", edgecolor='k', s=100)
plt.title('PCA of Customer Data')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True)
plt.show()

```
